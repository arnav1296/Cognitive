## Q1. Create the dataset

The dataset shown in the assignment contains 10 records with the columns:
`Refund`, `Marital Status`, `Taxable Income`, and `Cheat`.


```python
import pandas as pd
import numpy as np

data = {
    "Refund": ["Yes", "No", "No", "Yes", "No", "No", "Yes", "No", "No", "No"],
    "Marital Status": ["Single", "Married", "Single", "Married", "Divorced",
                       "Married", "Divorced", "Single", "Married", "Single"],
    "Taxable Income": ["125K", "100K", "70K", "120K", "95K",
                       "60K", "220K", "85K", "75K", "90K"],
    "Cheat": ["No", "No", "No", "No", "Yes", "No", "No", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)
print(df)
```

## Q2. Locate rows 0, 4, 7 and 8

Using `iloc`, which uses zero-based integer positions.


```python
rows = df.iloc[[0, 4, 7, 8]]
print(rows)
```

## Q3. Navigate the DataFrame

### 1. Select rows from index 3 to 7
### 2. Select rows from index 4 to 8 and columns 2 to 4
### 3. Select all rows with column index 1 to 3 (including column 3)

`iloc` uses an exclusive stop index, so the stop position is one greater than the last position we want.


```python
# 1. Rows from index 3 to 7 (inclusive)
q3_1 = df.iloc[3:8]
print("1. Rows 3 to 7:")
print(q3_1)

# 2. Rows 4 to 8 and columns 2 to 4 (inclusive)
q3_2 = df.iloc[4:9, 2:5]
print("\n2. Rows 4 to 8, columns 2 to 4:")
print(q3_2)

# 3. All rows, columns 1 to 3 (inclusive)
q3_3 = df.iloc[:, 1:4]
print("\n3. All rows, columns 1 to 3:")
print(q3_3)
```

## Q6. Create `employees.csv`

The assignment gives the following five employees and asks us to perform several Pandas operations on this dataset.


```python
employees_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employees_data)
employees["Joining_Date"] = pd.to_datetime(employees["Joining_Date"])

# Save the original sample dataset
employees.to_csv("employees.csv", index=False)

print(employees)
```

### Q6(e). Calculate statistics

1. Average salary
2. Total bonus
3. Youngest employee's age
4. Highest performance rating


```python
average_salary = employees["Salary"].mean()
total_bonus = employees["Bonus"].sum()
youngest_age = employees["Age"].min()
highest_rating = employees["Rating"].max()

print("Average salary:", average_salary)
print("Total bonus:", total_bonus)
print("Youngest employee's age:", youngest_age)
print("Highest performance rating:", highest_rating)
```

### Q6(f). Sort by Salary in descending order


```python
sorted_employees = employees.sort_values(by="Salary", ascending=False)
print(sorted_employees)
```

### Q6(g). Add a performance category

- **Excellent** for ratings >= 4.5
- **Good** for ratings >= 4.0 and < 4.5
- **Average** for ratings < 4.0


```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(performance_category)

print(employees[["Name", "Rating", "Performance_Category"]])
```

### Q6(h). Identify missing values


```python
print(employees.isnull().sum())
```

### Q6(i). Rename `Employee_ID` to `ID`


```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
print(employees)
```

### Q6(j). Find employees with more than 5 years of experience and employees in IT


```python
more_than_5_years = employees[employees["Years_of_Experience"] > 5]
it_employees = employees[employees["Department"] == "IT"]

print("Employees with more than 5 years of experience:")
print(more_than_5_years)

print("\nEmployees belonging to the IT department:")
print(it_employees)
```

### Q6(k). Add a `Tax` column

The Tax column deducts 10% of the Salary.


```python
employees["Tax"] = employees["Salary"] * 0.10

print(employees[["ID", "Name", "Salary", "Tax"]])
```

### Q6(l). Save the modified DataFrame to a new CSV file


```python
employees.to_csv("employees_modified.csv", index=False)

print("Modified DataFrame saved as employees_modified.csv")
print(employees)
```
