```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a
print("a.", temperature + 2)

# b
print("b.", (9/5) * temperature + 32)

# c
print("c.", temperature[temperature > 32])

# d
print("d.", np.sum(temperature > 32))

# e
print("e. Vectorization runs in compiled C code on whole arrays at once, avoiding slow Python loops.")
print("   Boolean indexing filters data in a single expression, so code is shorter, faster and cleaner.")
```

Q2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])

# a
print("a. Total:", steps.sum())

# b
print("b. Mean:", steps.mean())

# c
print("c. Max:", steps.max(), " Min:", steps.min())

# d
print("d. Per day:", steps.sum(axis=0))

# e
print("e. Per user:", steps.sum(axis=1))

# f
idx = steps.argmax()
print("f. Flat index:", idx, " Position (row, col):", np.unravel_index(idx, steps.shape))
```

Q3

```python
# a
original = np.array([1, 2, 3, 4, 5, 6])
print("a.", original)

# b
subset = original[1:5]
print("b.", subset)

# c
subset[0] = 999
print("c. original:", original)
print("   subset:", subset)

# d
copied = original[1:5].copy()
copied[0] = 500
print("d. original:", original)
print("   copied:", copied)

# e
matrix = np.arange(1, 13).reshape(3, 4)
print("e.\n", matrix)

# f
print("f. First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Rows 1-2, cols 2-3:\n", matrix[0:2, 1:3])

# g
flat = matrix.flatten()
rav = matrix.ravel()
print("g. flatten:", flat)
print("   ravel:", rav)

# h
rav[0] = 100
print("h. Original after changing ravel:\n", matrix)
print("   Original affected: Yes (ravel returns a view)")

# i
flat[1] = 200
print("i. Original after changing flatten:\n", matrix)
print("   Original affected: No (flatten returns a copy)")

# j
print("j. shape:", matrix.shape, " ndim:", matrix.ndim, " size:", matrix.size, " dtype:", matrix.dtype)
```

Q4

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a
print("a. shape:", X.shape, " ndim:", X.ndim)

# b
print("b. X.T:\n", X.T)
print("   Transpose swaps rows and columns: features become rows, users become columns.")

# c
XtX = X.T @ X
print("c. X.T @ X:\n", XtX)

# d
XtX_inv = np.linalg.inv(XtX)
print("d. Inverse:\n", XtX_inv)

# e
beta = XtX_inv @ X.T @ y
print("e. beta:", beta)

# f
print("f. Coefficients = change in assistance score per unit increase in")
print("   sleep hours, activity level and stress level respectively (others held fixed).")

# g
new_user = np.array([5, 40, 7])
print("g. Predicted score:", new_user @ beta)
```
