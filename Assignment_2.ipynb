```python
roll = "1024170251"
L = []
for i in roll:
    digit = int (i)
    L.append(digit*10)

print(L)
    
```

```python
L.append(100)
print(L)
```

```python
L.insert(1, 5123)
print(L)
```

```python
L.remove(5123)
print(L)
```

```python
L.pop() # pop removes last ele
print(L)
```

```python
L.sort()
print(L)
L.sort(reverse = True)
print(L)
```

```python
print(L[:3])
print(L[-3:])
```

```python
avg = sum(L)/len(L)
print(avg)
l2 = [i for i in L if i > avg]
print(l2)
```

```python
scores = tuple(L[0:8])
print(scores)
```

```python
highest = max(scores)
high_idx = scores.index(highest)
print("highest val: ", highest, "highest index: ", high_idx) 
lowest = min(scores)
low_cnt = scores.count(lowest)
print("lowest val: ", lowest, "count=", low_cnt)
```

```python
rev_tuple = list(scores[::-1]) # tuples cannot be mutatesd
print(rev_tuple)
```

```python
score = int(input("Enter a score: "))
if score in scores:
    print("Found at idx: ", scores.index(score))
else:
    print("Not found")
    
```
