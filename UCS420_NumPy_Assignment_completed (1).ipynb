```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(arr + 2)
print(arr * 3)
print(arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print(arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print(arr[0, 1])
print(arr[2, 0])
```

```python
Arnav = np.linspace(10, 100, 25)

print("Dimensions:", Arnav.ndim)
print("Shape:", Arnav.shape)
print("Total elements:", Arnav.size)
print("Data type:", Arnav.dtype)
print("Total bytes:", Arnav.nbytes)

transpose_Arnav = Arnav.reshape(1, 25).T
print(transpose_Arnav)

print(Arnav.T)
```

```python
ucs420_Arnav = np.array([[10, 20, 30, 40],
                        [50, 60, 70, 80],
                        [90, 15, 20, 35]])

print("Mean:", np.mean(ucs420_Arnav))
print("Median:", np.median(ucs420_Arnav))
print("Max:", np.max(ucs420_Arnav))
print("Min:", np.min(ucs420_Arnav))
print("Unique elements:", np.unique(ucs420_Arnav))

reshaped_ucs420_Arnav = ucs420_Arnav.reshape(4, 3)
print(reshaped_ucs420_Arnav)

resized_ucs420_Arnav = np.resize(ucs420_Arnav, (2, 3))
print(resized_ucs420_Arnav)
```
