```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_temperature = temperature + 2
print(corrected_temperature)

fahrenheit = 9/5 * temperature + 32
print(fahrenheit)

greater_than_32 = temperature[temperature > 32]
print(greater_than_32)

count = np.sum(temperature > 32)
print(count)

print("Vectorization applies operations to the entire array at once, reducing explicit iteration. Boolean indexing directly selects values satisfying a condition, making array processing efficient.")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(steps.sum())
print(steps.mean())
print(steps.max())
print(steps.min())
print(steps.sum(axis=0))
print(steps.sum(axis=1))

max_index = np.argmax(steps)
print(np.unravel_index(max_index, steps.shape))
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999
print(original)
print(subset)

copied_array = original[1:5].copy()
copied_array[0] = 500
print(original)
print(copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

print(matrix[0])
print(matrix[-1])
print(matrix[:, 1])
print(matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()
print(flattened)
print(raveled)

raveled[0] = 100
print(matrix)
print(raveled)

flattened[1] = 200
print(matrix)
print(flattened)

print(matrix.shape)
print(matrix.ndim)
print(matrix.size)
print(matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print(X.shape)
print(X.ndim)

print(X.T)

XtX = X.T @ X
print(XtX)

XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)

beta = XtX_inv @ X.T @ y
print(beta)

print("The coefficients represent the change in predicted assistance score associated with a one-unit change in sleep hours, activity level, and stress level, respectively, while the other features are held constant.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(predicted_score)
```
